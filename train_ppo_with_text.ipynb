# PPO With Text Features

Portable notebook wrapper for the Toy Example pipeline. Run it from an environment with FinRL, Stable-Baselines3, PyTorch, pandas, numpy, and pyfolio.

```python
from pathlib import Path
import subprocess
import sys

ROOT = Path.cwd()
if not (ROOT / 'scripts').exists():
    ROOT = Path(__file__).resolve().parent if '__file__' in globals() else ROOT

PANEL = ROOT / 'artifacts' / 'processed_final_fixed_external_lagclean_full_WITH_TEXT_MISTRAL.csv'
OUT = ROOT / 'artifacts' / 'ppo_with_text_run'

print('Python:', sys.executable)
print('Root:', ROOT)
print('Panel exists:', PANEL.exists())
```

## 1. Validate inputs

```python
subprocess.run([
    sys.executable,
    'scripts/03_train_backtest_ppo_with_text.py',
    '--panel', str(PANEL),
    '--output-dir', str(OUT),
    '--dry-run',
], cwd=ROOT, check=True)
```

## 2. Train or resume PPO with text

```python
RESUME = True

cmd = [
    sys.executable,
    'scripts/03_train_backtest_ppo_with_text.py',
    '--panel', str(PANEL),
    '--output-dir', str(OUT),
    '--timesteps', '350000',
    '--save-freq', '50000',
]
if RESUME:
    cmd += ['--resume-from-checkpoint', 'latest', '--force-train']

subprocess.run(cmd, cwd=ROOT, check=True)
```

## 3. Compare against benchmark

```python
subprocess.run([
    sys.executable,
    'scripts/04_compare_with_benchmark.py',
], cwd=ROOT, check=True)

print((ROOT / 'artifacts' / 'ppo_text_vs_benchmark' / 'report.md').read_text(encoding='utf-8'))
```
